
````python
"""Devuelve el mayor de tres números sin usar max()."""

def mayor_de_tres(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

print("Resultado para mayor_de_tres(4, 9, 2):")
print(mayor_de_tres(4, 9, 2))
````

EJERCICIO 5
```
# Tiene formato de código
Algoritmo: Suma de pares entre 1 y n
Entrada: n
salida: suma
bloque de desarrollo:
    suma = 0
    PARA i DESDE 1 HASTA n:
        SI i % 2 == 0:
            suma = suma + i
    DEVOLVER suma
```

````python
"""Suma los números pares entre 1 y n."""

def suma_pares(n):
    suma = 0
    for i in range(1, n + 1):
        if i % 2 == 0:
            suma += i
    return suma


print("Resultado para suma_pares(10):")
print(suma_pares(10))
````

EJERCICIO 6
```
# Tiene formato de código
Algoritmo: Factorial con bucle
Entrada: n
salida: resultado
bloque de desarrollo:
    resultado = 1
    PARA i DESDE 1 HASTA n:
        resultado = resultado * i
    DEVOLVER resultado
```

````python
"""Calcula n! usando un bucle (versión iterativa)."""

def factorial_iterativo(n):
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    return resultado

print("Resultado para factorial_iterativo(5):")
print(factorial_iterativo(5))
````

```
# Tiene formato de código
Algoritmo: Contar vocales de una palabra
Entrada: palabra
salida: contador
bloque de desarrollo:
    contador = 0
    vocales = "aeiou"
    PARA cada letra EN palabra en minusculas:
        SI letra ESTA EN vocales:
            contador = contador + 1
    DEVOLVER contador
```

````python
"""Cuenta las vocales de una palabra, sin distinguir mayúsculas."""

def contar_vocales(palabra):
    contador = 0
    vocales = "aeiouAEIOU"
    for letra in palabra:
        if letra in vocales:
            contador += 1
    return contador

print("Resultado para contar_vocales('Programacion'):")
print(contar_vocales('Programacion'))
````

```
# Tiene formato de código
Algoritmo: Promedio de una lista de números
Entrada: numeros (lista)

    SI la lista esta vacia:
        DEVOLVER 0.0
    suma = suma de todos los elementos
    promedio = suma / longitud de numeros
```

````python
"""Calcula la media aritmética de una lista de números."""

def calcular_promedio(numeros):
    if not numeros:
        return 0.0
    return sum(numeros) / len(numeros)

print("Resultado para calcular_promedio([4, 8, 6]):")
print(calcular_promedio([4, 8, 6]))
````

```

Algoritmo: Texto en mayúsculas y número de palabras
Entrada: frase
salida: frase_mayusculas, num_palabras

    frase_mayusculas = frase en mayusculas
    palabras = separar frase por espacios

    DEVOLVER frase_mayusculas, num_palabras
```

````python
"""Devuelve la frase en mayúsculas junto con su número de palabras."""

def analizar_frase(frase):
    frase_mayusculas = frase.upper()
    num_palabras = len(frase.split())
    return frase_mayusculas, num_palabras

print("Resultado para analizar_frase('hola que tal'):")
print(analizar_frase('hola que tal'))
````

```
# Tiene formato de código
Algoritmo: Conversión de temperatura
Entrada: celsius / fahrenheit
salida: fahrenheit / celsius
bloque de desarrollo:
    Para Celsius a Fahrenheit: DEVOLVER (celsius * 9/5) + 32
    Para Fahrenheit a Celsius: DEVOLVER (fahrenheit - 32) * 5/9
```



````python
"""Convierte una temperatura de Celsius a Fahrenheit."""

def celsius_a_fahrenheit(celsius):
    return (celsius * 9 / 5) + 32

"""Convierte una temperatura de Fahrenheit a Celsius."""

def fahrenheit_a_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print("Resultado para celsius_a_fahrenheit(0):")
print(celsius_a_fahrenheit(0))

print("Resultado para fahrenheit_a_celsius(32):")
print(fahrenheit_a_celsius(32))
````

# **FUNCIONES RECURSIVAS**

EJERCICIO 13


```
# Tiene formato de código
Algoritmo: Factorial recursivo
Entrada: n
salida: resultado
bloque de desarrollo:
    SI n == 0 O n == 1:
        DEVOLVER 1
    SI NO:
        DEVOLVER n * factorial_recursivo(n - 1)
```



````python
"""Calcula n! llamándose a sí misma (versión recursiva)."""

def factorial_recursivo(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursivo(n - 1)

print("Resultado para factorial_recursivo(5):")
print(factorial_recursivo(5))
````

EJERCICIO 14


```
# Tiene formato de código
Algoritmo: Sucesión de Fibonacci
Entrada: n
salida: termino_n
bloque de desarrollo:
    SI n == 0:
        DEVOLVER 0
    SI NO SI n == 1:
        DEVOLVER 1
    SI NO:
        DEVOLVER fibonacci(n - 1) + fibonacci(n - 2)
```



````python
"""Devuelve el término n de la sucesión de Fibonacci."""

def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

print("Resultado para fibonacci(6):")
print(fibonacci(6))
````

```
# Tiene formato de código
Algoritmo: Suma recursiva de una lista
Entrada: numeros (lista)
salida: suma
bloque de desarrollo:
    SI la lista esta vacia:
        DEVOLVER 0
    SI NO:
        DEVOLVER primer_elemento + suma_lista(resto_de_la_lista)
```



````python
"""Suma los elementos de una lista de forma recursiva."""

def suma_lista(numeros):
    if not numeros:
        return 0
    return numeros[0] + suma_lista(numeros[1:])

print("Resultado para suma_lista([1, 2, 3, 4]):")
print(suma_lista([1, 2, 3, 4]))
````

EJERCICIO 16


```
# Tiene formato de código
Algoritmo: Invertir una cadena de texto
Entrada: texto
salida: texto_invertido
bloque de desarrollo:
    SI longitud de texto <= 1:
        DEVOLVER texto
    SI NO:
        DEVOLVER ultimo_caracter + invertir_texto(resto_del_texto)
```



````python
"""Invierte una cadena de texto de forma recursiva."""

def invertir_texto(texto):
    if len(texto) <= 1:
        return texto
    return texto[-1] + invertir_texto(texto[:-1])

print("Resultado para invertir_texto('python'):")
print(invertir_texto('python'))
````

```
# Tiene formato de código
Algoritmo: Calculadora con control de errores
Entrada: texto_a, texto_b, operacion
salida: resultado o mensaje de error
bloque de desarrollo:
    INTENTAR:
        num_a = float(texto_a)
        num_b = float(texto_b)
    EN CASO DE ERROR DE VALOR:
        DEVOLVER "Error: valor no numérico"

    SI operacion == "suma":
        DEVOLVER num_a + num_b
    SI NO SI operacion == "division":
        INTENTAR:
            DEVOLVER num_a / num_b
        EN CASO DE DIVISION POR CERO:
            DEVOLVER "Error: división entre cero"
```



````python
"""Suma o divide dos textos numéricos controlando los errores posibles."""

def calculadora_segura(texto_a, texto_b, operacion):
    try:
        num_a = float(texto_a)
        num_b = float(texto_b)
    except ValueError:
        return "Error: valor no numérico"

    if operacion == "suma":
        return num_a + num_b
    elif operacion == "division":
        try:
            return num_a / num_b
        except ZeroDivisionError:
            return "Error: división entre cero"

print("Resultado para calculadora_segura('10', '0', 'division'):")
print(calculadora_segura('10', '0', 'division'))
````
